```python
import matplotlib.pyplot as plt
import numpy as np

# Set random seed for reproducibility
np.random.seed(42)


def buffon_needle_sim(
    num_needles=1000000, line_spacing=1.0, needle_length=0.8
):
    """Buffon's Needle Monte Carlo Simulation to Estimate Pi."""
    # 1. Generate random uniform variables
    # x: Distance from needle midpoint to closest line, uniform in [0, a/2]
    x = np.random.uniform(0, line_spacing / 2.0, num_needles)

    # theta: Angle between needle and parallel lines, uniform in [0, pi/2]
    theta = np.random.uniform(0, np.pi / 2.0, num_needles)

    # 2. Check intersection condition: x <= (l/2) * sin(theta)
    hits = x <= (needle_length / 2.0) * np.sin(theta)

    # 3. Calculate estimation
    hit_count = np.sum(hits)
    pi_estimate = (2 * needle_length * num_needles) / (
        line_spacing * hit_count
    )

    print(f"Total Needles (N): {num_needles}")
    print(f"Intersections (n): {hit_count}")
    print(f"Estimated Pi:      {pi_estimate:.6f}")
    print(f"Absolute Error:    {abs(pi_estimate - np.pi):.6f}")

    return x, theta, hits, pi_estimate


# Execute simulation
N = 500000
x, theta, hits, pi_est = buffon_needle_sim(num_needles=N)
```
