1733年，法国皇家科学院的报告中写道：This year M. le Clerc de Buffon presented to the Academy some solutions of problems which looked into the game of franc-Carreau. 

Buffon正在研究Franc-Carreau游戏——一枚圆形硬币抛出后恰好落在一块瓷砖之内的概率有多大？

The question presented is of a new kind, to those which belong to Geometry, & to the figures which were not yet entered into this matter，当时人们觉得这是一类未被定义的几何问题。

硬币相对于单块等大砖块的比例越小，押注硬币完全落在砖块内的胜率就越高。Buffon观察到圆形硬币不压线取决于圆心位置，所以他在大正方形A里内嵌了一个小正方形B，当圆心在B内时，硬币就不会压线。大正方形A的边长为a,圆形硬币的半径为r。

![Franc-Carreau 几何抽象示意图](assets/1.png)

18世纪的概率刚从赌博问题开始，所以Buffon讨论了赌博的问题——一个人压圆形硬币落在砖内，另一个人压圆形硬币压到砖缝，什么情况下二人获胜是公平的？即二人获胜的概率相同，即正方形A的面积 和 A与B之间的边框面积 相等。

$$(a-2r)^2 = a^2 - (a-2r)^2$$
$$\frac{r}{a} \approx 0.1464 < \frac{1}{6}$$

所以a要大于6倍的r，这样游戏才公平，Buffon抛出另一个问题：如果是方形硬币，怎么办？这里面掺杂着复杂的几何运动，这份报告里没有展开讨论，而是笔锋一转，提出*在仅由等宽且平行的木板铺成的平面上，人们抛掷一根具有一定长度且假设没有宽度的针（细棒）。这根针何时会不压到砖缝？*，这就是我们今天看到的buffon投针问题。

把圆形硬币换作针，题目发生了什么变化？

圆形硬币的位置确定是依靠圆心的，那么针呢？除了它的中心，还有一个关键问题，那就是针与木板边之间的角度——增加了一个自由度。

那么我们看看这份报告里如何讨论这个问题。


首先，当针与木板的边缘平行时，必然会完全落在单块木板。

假设针的中心点距离木板边缘有一定距离，以针中心为圆心，通过距离该边缘最近的针端点画一个四分之一圆弧，该圆弧的一部分会在木板内部，另一部分则在木板外部。针在木板内部有多少种姿态，就会对应多少次“完全落入”；反之亦然。

因此，“完全落入”的次数与跨线次数之比，等于针在内部姿态的总和与外部姿态总和之比；或者说，等于该四分之一圆被木板边缘分隔开的内部与外部两部分面积之比。

显然，在解决这一整体问题时，必须考量针的长度与木板宽度的比值。

1.若两者相等，则仅当木棍中心点落于木板宽度的中心点时，木棍方能完全落在木板之上；2.若木板宽度大于针的长度，木棍中心点可落入的有效位置点数便随之增多。

因此，必然存在某种特定的木板宽度，能使该赌局或游戏处于均势；勒·克莱尔先生（Buffon）正是利用摆线的面积（针扫过的面积），以一种极具巧思的方式确定了这一宽度，并将其呈报给科学院审议。

![法国科学院](assets/2.png)

以上思考，成功地把一个概率问题转化为了几何问题，这就是Buffon对于概率的创造性体现。

为什么这里用的是$\frac{1}{4}$圆？为什么没有使用π？

第一个问题，这里是因为针的中心点确定后，针可以 $360^\circ$ 任意旋转。但因为针是两头对称的，研究 $180^\circ$ 和研究 $360^\circ$ 效果完全一样。再考虑与平行木板缝隙的垂直对称性（向左偏和向右偏到木板缝隙的距离是一致的），针的端点距离缝隙最近的可能角度正好分布在一个象限内，即 $0^\circ$ 到 $90^\circ$。因此，布丰只需要在这个四分之一圆（$90^\circ$ 扇形）里计算“针落入板内”的角度弧长/扇形面积占整个四分之一圆的比例即可。

第二个问题，是因为当时的微积分刚出现，数学语言比较模糊，数学家更习惯用面积和弧长来研究，而非用π表示。

Buffon在1777年的著作《道德算术散论》的XXIII.部分，谈到了投针问题，他在书中写道：Analysis is the only instrument by which one is served until this day in the science of Geometric probability probabilities, to determine & to fix the ratios of risk; Geometry appeared ill-suited to a work so delicate; however if one considers it closely, it will be easy to recognize that this advantage of Analysis on Geometry, is completely accidental, & that risk according as it is modified & composed, is found as a result of geometry as well as that of analysis; in order to be assured of it, it will be sufficient to pay attention that the games & the questions of conjecture turn customarily only on the ratios of discrete quantities; the human spirit more familiar with numbers than with measures of extent have always preferred them; the games are one proof of it, because their laws are one continual arithmetic;（译文：在几何概率这一学科中，迄今为止，分析法一直是确定和设定风险比率的唯一工具；乍看之下，几何法似乎并不适合处理如此精细的问题；然而，若仔细考量，便不难发现分析法相对于几何法的这种优势纯属偶然——事实上，无论风险的形式如何变化或组合，既可以通过几何法，也可以通过分析法得出结果；要确信这一点，只需留意博弈与概率问题通常仅涉及离散量的比率即可；人类思维因更熟悉数字而非度量尺度，故而总是偏好数字；各类博弈便是明证，因为其规则本质上就是连续不断的算术运算。）

我觉得这一段写的太具有划时代意义了，所以特地摘出来，他指出过去的概率计算（费马、帕斯卡等人研究的概率论）总是基于数字分析，而不是几何，针对这个现象，他归因于“人类思维因更熟悉数字而非度量尺度，故而总是偏好数字”，在我们今天的视角，人尽皆知古典概型是面对局限的离散量，而不是连续，是几何概型第一次让人类对于概率的思考走到了连续，而Buffon先生就是几何概型的先驱，他在当年并不是懵懵懂懂地提出了投针问题，而是很有前瞻性、预见性地指出人类对于概率问题研究的局限。

But if instead of throwing in the air a round piece, as a coin, one will throw a piece of
another figure, as a square pistole of Spain, or a needle, a rod, &c. the problem will demand
a little more geometry, although in general it is possible always to give the solution by the
comparison of spaces, as we are going to demonstrate.
I suppose that in a room, of which the floor is simply divided by parallel joints, one 
throws into the air a rod, & that one of the players wagers that the rod will not cross any
of the parallels of the floor, and that the other to the contrary wagers that the rod will cross
some one of the parallels; one requires the lot of the two players. One is able to play this
game on a draught board with a sewing needle or a headless pin.（译文：然而，如果抛向空中的不是像硬币那样的圆形物体，而是其他形状的物体——例如西班牙方形金币、缝衣针或细杆等——那么这个问题就需要运用更多的几何知识来解决；不过，正如我们将要证明的那样，通常总能通过比较面积来得出答案。

假设在一个地板上仅由平行线划分的房间里，有人将一根细杆抛向空中；其中一名参与者打赌这根细杆不会跨越地板上的任何平行线，而另一名参与者则打赌它会跨越其中某条平行线。现在的任务是确定这两名参与者各自的胜算。这个游戏也可以在棋盘上用缝衣针或无头大头针来进行。）

‘为求得此概率，我首先在地面两条平行的接缝 $AB$ 与 $CD$ 之间，画出另外两条平行线 $ab$ 与 $cd$；

这两条新平行线分别与原接缝保持一段距离，距离恰等于细木条 $EF$ 长度的一半。

我清楚地看到：只要细木条的中点处于ab和cd之间，那么无论细木条处于何种位置，它都绝不可能跨越原有的接缝。又因为在 $ab$ 上方可能发生的一切情况，都完全对称地发生在 $cd$ 下方，因此我们只需要确定其中一种情况即可。

为此，我注意到：细木条的所有可能旋转位置，都可以用一个以细木条长度为直径的圆周的四分之一来表示。

因此，设地板接缝 $CA$ 的间距为 $2a$，细木条的长度为 $2b$，以细木条长度2b为直径的圆的周长四分之一为 $c$，接缝的长度 $AB$ 为 $f$；我将得到 $(a - b)fc$ 这个表达式(在那个测度不明确的年代，没有使用今天的面积来刻画样本空间，而是将三个量相乘来表示，这个式子没有现实意义，后面要作比，所以可以用，不要纠结)，它表示细木条不跨越地板接缝的概率空间；换句话说，这代表了细木条中点落入 $ab$ 线下方且处于 $cd$ 线上方的所有情形。’

![布丰投针几何图](assets/3.png)

但是，当细木条的中点落入区域 $abdc$ 之外时，根据其所处的位置，细木条既可能跨越接缝，也可能不跨越。

因此，假设细木条的中点位于 $\epsilon$ 处，圆弧 $\phi G$ 将代表细木条会跨越接缝的所有位置，而圆弧 $GH$ 则代表不会跨越的所有位置。

由于对于线段 $\epsilon\phi$ 上的所有点情况皆是如此，我将这条线段上的微小微元称为 $dx$（微元是指木条中点到接缝的垂直距离线段上的微元），将圆弧 $\phi G$ 称为 $y$。

于是，我得到 $\left(\int y \, dx\right)f$ 作为细木条会跨越接缝的所有情形的表达式(表示跨线出去的一部分，y是c在缝隙外的部分圆弧，dx是线段b上的微元)；而对于不会跨越的情形，表达式则为 $\left(bc - \int y \, dx\right)f$（b表示半径，从表示圆弧，所以你可以看到三个量一直围绕半径b、圆弧长c和缝隙长度f，我觉得f是多余的，反而容易让人混淆）。

我将后一个表达式与上文求得的 $(a - b)fc$ 相加，从而得到细木条不会跨越接缝的所有情形的总和。

由此可见，第一位玩家（下注不跨越）与第二位玩家（下注跨越）的获胜概率之比为：$$ac - \int y \, dx : \int y \, dx$$因此，如果希望游戏保持公平，就必须满足 $ac = 2 \int y \, dx$，即 $a = \frac{\int y \, dx}{\frac{1}{2}c}$；也就是说，这等于生成圆直径为细木条长度 $2b$ 的摆线（Cycloid）的一部分面积。

现已知该摆线的面积等于半径的平方$b^2$（这里说的摆线不是我们理解的摆线，应该是四分之一圆在缝隙外的部分），因此有 $a = \frac{bb}{\frac{1}{2}c}$。也就是说，若要使游戏公平，细木条的长度必须设为地板接缝间距的约四分之三（化简完是π/4）。

这第一种情形的求解，引领我们轻松地得出另一种乍看之下更为困难的情形——即确定在用正方形瓷砖铺设的房间里，这两位玩家的获胜概率。

因为通过在其中一块正方形瓷砖内内接一个各边均距原正方形边长为 $b$ 的正方形（这里的内接也不是我们理解的内接，而是把一个边长为$a$的正方形整体缩小为边长为$a-b$的正方形，中心不变），我们首先可以得到 $c(a - b)^2$，作为细木条不跨越任何接缝的部分情形的表达式（这里依然是用三个量相乘刻画空间，a-b是边长，c是圆弧长，代表角度的维度，加下来的过程和上面差不多，只是把一个量变味了a-b）；

接着，我们将求得 $(2a - b) \int y \, dx$，作为细木条会跨越接缝的所有情形的表达式；

最后，得到 $cb(2a - b) - (2a - b) \int y \, dx$，作为细木条不跨越接缝的其余情形的表达式。

因此，第一位玩家（下注不跨越）与第二位玩家（下注跨越）的获胜概率之比为：$$c(a - b)^2 + cb(2a - b) - (ca - b) \int y \, dx : (2a - b) \int y \, dx$$

因此，如果希望游戏保持公平，就必须满足 $c(a - b)^2 + cb(2a - b) = (2a - b)^2 \int y \, dx$ ，即：$$\frac{\frac{1}{2} c a^2}{2a - b} = \int y \, dx$$

但是，正如我们上文所见，$\int y \, dx = b^2$ ；因此有：$$\frac{\frac{1}{2} c a^2}{2a - b} = b^2$$由此可得，正方形瓷砖的边长与细木条长度之比，大约应当为 $\frac{41}{22} : 1$，也就是说，不到两倍（约 1.86 倍）。

因此，如果有人在棋盘上玩投针游戏，且针的长度刚好等于棋盘正方形格边长的一半，那么下注“针会跨越接缝”的玩家将占据优势。通过类似的计算人们还会发现：如果是用一枚方形硬币来进行游戏，所有情况的总和与下注“压缝”玩家的获胜概率之比，将为：$$a^2 c : 4a b^2 \sqrt{\frac{1}{2}} - b^3 - \frac{1}{2} A b$$其中，$A$ 在此处表示该正方形外接圆的面积与正方形自身面积的差值（即外接圆超出正方形的面积部分），而 $b$ 表示该正方形的半对角线长。

在最后，他写道These examples suffice to give an idea of the games that one is able to imagine on the
ratios of area ......where one must consider that the ratio of the area, & which by consequence concern Geometry all as much as Analysis.(翻译：这些例子足以让人领略到，人们可以围绕面积比构想出怎样的游戏——毕竟，面积比的问题既关乎几何，也同样关乎分析。)

接下来是1812年，Laplace在《概率分析论》里面的讨论：

我们可以利用概率计算来进行曲线的求弧长或求其表面积（求积分）。毫无疑问，几何学家们一般不会采用这种方法；但由于它为我提供了一个探讨特殊类型偶然组合的机会，我将用简短的语言对其进行阐述：

设想一个平面被一组间距为 $a$ 的平行线所分割；再设想一个极细的圆柱体（针），其长度为 $2r$，假设$2r$等于或小于 $a$。现要求求解抛掷该圆柱体时，它与平面上某条平行线相交的概率。

在这些分割线的某一点上作一条垂直线，延伸至下一条平行线。假设圆柱体的中心位于这条垂直线上，且在第一条平行线上方高度为 $y$ 处。让圆柱体绕其中心旋转，设 $\phi$ 为圆柱体与该垂直线所成的夹角（在圆柱体恰好与平行线相交的时刻）。那么，$2\phi$ 将代表圆柱体端点所扫过圆周中能使其与平行线相交的那部分弧长(当年对于角度的定义就是单位圆所对应的弧长，这里的弧长不是我们理解的弧长，所以少了半径)；因此，所有这些弧长的总和即为 $4\int \phi \, dy$ 或 $4\phi y - 4\int y \, d\phi$（这里的弧长总和说的不是弧长，而是弧长扫过的面积，$2\phi$的角度积累 $2r\phi$的弧长，这里虽然没有写r，但是后面在对$y$的定义里补上了）；又因为我们有 $y = r \cos\phi$，因此该总和等于：$$4\phi y - 4r \sin\phi + \text{常数}$$

![](assets/4.png)

为了确定这个常数，我们注意到积分必须从 $y = 0$ 延伸到 $y = r$，因此相应地从 $\phi = \frac{\pi}{2}$ 延伸到 $\phi = 0$，这给出：$$\text{常数} = 4r$$因此，我们所讨论的总和为 $4r$。从 $y = a - r$ 到 $y = a$，圆柱体（针）可以与下一条分割线相交，显而易见，与该相交相关的各部分之和同样为 $4r$。因此，当圆柱体的中心沿垂直线移动时，与圆柱体相交于两条分割线中任意一条相关的总和为 $8r$。但是，圆柱体在垂直线的每个点上绕自身完整旋转所扫过的全部弧长之和为 $2a\pi$；这就是所有可能组合的总数。因此，圆柱体与平面上某条分割线相交的概率为：$$\frac{4r}{a\pi}$$如果我们将该圆柱体抛掷多次，根据第16号定理(当独立重复试验的次数 $n$ 非常大时，事件发生的实际频率（相交次数 / 抛掷总次数）会以极高的概率无限接近其理论概率 $P$)，圆柱体与平面分割线相交的次数与抛掷总次数的比值，将非常精确地接近 $\frac{4r}{a\pi}$ 的值，由此可以求得圆周率 $2\pi$ 的值。根据同一号定理，我们还可以求出该值的误差保持在给定范围内的概率；并且很容易看出，对于给定的抛掷次数，使需担心的误差达到最小的比值为 $\frac{8r}{a\pi} = 1$（即总概率为1时），这给出的圆柱体长度等于分割线的间距乘以圆周长与四个直径的比值。现在，我们设想上述平面被另一组与前者垂直、且间距为 $b$（ $b$ 大于或等于圆柱体长度 $2r$）的平行线所分割。所有这些线将与第一组线共同构成一系列长为 $b$、高为 $a$ 的矩形。

考虑其中一个矩形：假设在其内部，与各边平行的方向上分别保持距离 $r$ 作平行线。它们首先将形成一个矩形，其长为 $b - 2r$，高为 $a - 2r$；其次是两个小矩形，高为 $r$，长为 $b - 2r$；接着是另外两个小矩形，长为 $r$，高为 $a - 2r$；最后是四个边长等于 $r$ 的小正方形。(这里回到矩形讨论，前面的Buffon也是在平行线之后讨论网格形状，Buffon选择了正方形，而Laplace选择的是更一般的情形)

只要圆柱体（针）的中心处于这个内部矩形中，圆柱体绕其中心旋转时就绝不会与大矩形的各边相交。

当圆柱体的中心处于高为 $r$、长为 $b - 2r$ 的其中一个小矩形内部时，根据前文所述容易看出，$8r$ 乘以长度 $(b - 2r)$ 的乘积，即为圆柱体与大矩形两条长边 $b$ 中某一条相交的对应组合数。因此，$8r(b - 2r)$ 就是当圆柱体中心位于这两个小矩形中的某一个时，圆柱体与大矩形轮廓相交的所有对应组合的总数。

同理，$8r(a - 2r)$ 则是当圆柱体中心位于尺寸为 $r$ 和 $a - 2r$ 的小矩形内部时，圆柱体与大矩形轮廓相交的组合总数。

现在还剩下四个小正方形需要考虑。设其中一个是 $ABCD$。以该正方形与大矩形共用的顶点 $A$ 为圆心、以半径 $r$ 画弧，得到止于点 $B$ 和 $D$ 的四分之一圆弧。只要圆柱体的中心包含在该弧构成的四分之一圆内，圆柱体在旋转时，在其所有位置上都必定会与大矩形的轮廓相交。因此发生这种情况的组合数等于 $2\pi$ 乘以四分之一圆的面积，故其等于 $\frac{\pi^2 r^2}{2}$。如果圆柱体的中心位于该正方形中超出四分之一圆的部分，圆柱体绕其中心旋转时，可能会与 $AB$ 和 $AD$ 的延长线中的某一条相交，但绝不可能同时与两者相交。

![](assets/5.png)

为了确定与该相交相关的组合数，我设想在边 $AB$ 上距离 $A$ 点为 $x$ 的任意一点上，作一条垂线 $y$，使其端点位于四分之一圆之外。我将圆柱体（针）的中心置于该端点处，并从该点引出四条长等于 $r$ 的线段，其中两条止于延长后的 $AB$ 边（如有必要），另外两条止于同样延长后的 $AD$ 边。我将前两条线段所夹的角记为 $2\phi$，后两条线段所夹的角记为 $2\phi'$。显而易见，圆柱体绕其中心旋转时，只要其一半处于视角 $2\phi$ 内，就会与 $AB$ 的延长线相交；只要其一半处于视角 $2\phi'$ 内，就会与 $AD$ 的延长线相交。因此，圆柱体与这两条边中的某一条相交的组合总数为 $4(\phi + \phi')$；因此，对于正方形中位于四分之一圆外部的那部分区域，对应的组合数表示为：$$4 \iint (\phi + \phi') \, dx \, dy$$显然我们有：$$x = r \cos\phi', \quad y = r \cos\phi$$上述积分因而转化为：$$4r^2 \iint (\phi + \phi') \sin\phi \sin\phi' \, d\phi \, d\phi'$$很容易看出，关于 $\phi'$ 的积分范围应当从 $\phi' = 0$ 到 $\phi' = \frac{\pi}{2} - \phi$，而关于 $\phi$ 的积分范围应当从 $\phi = 0$ 到 $\phi = \frac{\pi}{2}$。这给出该积分的值为：$$\frac{1}{2} r^2 (12 - \pi^2)$$将其加上之前的 $\frac{\pi^2 r^2}{2}$，我们就得到了与整个小正方形相关的组合数。将这个数值乘以 4（四个顶角正方形），并将其与此前求得的与圆柱体跨越大矩形轮廓相关的组合数相加，我们最终得到总相交组合数为：$$8(a + b)r - 8r^2$$

但是，所有可能组合的总数显然等于 $2\pi$ 乘以大矩形的面积 $ab$；因此，圆柱体（针）与平面分割线相交的概率为：$$\frac{4(a + b)r - 4r^2}{ab\pi}$$

Laplace的计算过程比较容易理解，因为已经相对地靠近现代，他在1812年就提出可以用实验结果来倒推π的数值，这是很大胆的想法，也被视为后来montecarlo模拟的起点。

好了，现在到了最后，我不知道写些什么，又觉得总要说几句

不吐不快，我就随着性子聊几句

在我看完两位数学家关于一个问题的探讨后，我最大的感受可能不是发现了什么不可思议的东西，毕竟思路也还是围绕着（x,ɑ）做分析，和现代解法没什么区别，只不过是他们那个时候的数学语言并不像今天这么完善。

他们在那样一个萌芽的状态，就能够展现出惊艳的推导，很不可思议，我在想，是什么让他们能够在那样一个环境下完成这些事，可能是他们的天赋过人？很可能，但是除此之外呢？

我斗胆推测，他们对这个世界有强烈的好奇心。

一个今天看来甚至有些微不足道的问题，他们为什么还愿意花费那么多时间去思考？

至少在当时，这个问题并没有我们今天赋予它的那些意义。他们可能并不知道，这根针、这个圆、这个概率问题，后来会牵引出怎样漫长的数学与计算史。

也许他们只是单纯地觉得：

“这个问题很有意思，我想知道答案。”

他们可能就像牛顿所说的那样，只是一个孩子，偶尔在海边捡起一枚更加光滑、更加漂亮的贝壳，而他们面前的，是尚未被穷尽的真理海洋。

而这恰恰也是我做这个研究时最意外的收获。

刚梳理出来时，我感觉自己与伟大的灵魂对话，很兴奋，后来去看他们的计算过程时，属实困难，以致于我在深夜崩溃，我一度怀疑自己根本不懂数学，我感觉自己很奇怪，为什么要把这个翻出来，写出来，又有什么意义呢？

我不懂做这个有什么意义，因为解题的方法没什么变化，别人看不看都一样

直到今天，我向我的统计学老师分享的时候，我们聊到了我高中两次关于数学的挫折，竞赛和高考都因为身体突发状况而落败，说着说着，她叹了一口气，对我说“哎，你对数学真的很敏感”，我呆住了，反问“真的...吗？”，她说“是的，你现在即使在经济，也还是在做数学，而且研究得不浅”

我仿佛看到了高中那个对数学满怀热情的自己，主元思想、对称性......我当时在探寻背后的本质，而非追逐二级结论。

然而，在两次挫败里失去了信心，在转专业时，老师问我为什么不去数学院，我说我可能不适合

今天的统计学老师问我“为什么不来我们数学院啊”，我想自己可能不是“不适合”，只是我为了保护自己的伤口，而自我欺骗，甚至到最后，我自己都信了

如果只从“解题方法”来看，我甚至可以说：这一次，我好像什么都没有得到。但现在我却觉得，这可能恰恰是我得到的东西，因为我重新找到了那个曾经对数学感到好奇的人。那个不满足于知道“答案是多少”，而总想追问“为什么”的人。原来他一直都没有消失，只是我曾经因为那些失败，不敢再承认他的存在。

所以，当一位真正的专业学者告诉你：

“你对数学很敏感。”

你为什么还要假装不知道呢？
